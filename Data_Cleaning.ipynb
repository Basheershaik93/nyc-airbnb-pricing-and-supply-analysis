```python
working_df.columns = working_df.columns.str.lower().str.replace(' ', '_').str.strip()
```

```python
room_type_map ={
    'Entire home/apt':'Entire home'
}
working_df['room_type'] = working_df['room_type'].map(room_type_map).fillna(working_df['room_type'])
```

```python
working_df.head()
```

```python
print("Rows where both are null:", len(working_df[working_df['last_review'].isnull() & working_df['reviews_per_month'].isnull()]))
print("Only last_review null:", len(working_df[working_df['last_review'].isnull() & working_df['reviews_per_month'].notnull()]))
print("Only reviews_per_month null:", len(working_df[working_df['last_review'].notnull() & working_df['reviews_per_month'].isnull()]))
```

```python
working_df['reviews_per_month'] = working_df['reviews_per_month'].fillna(0)
working_df = working_df.drop('last_review', axis=1)
```
